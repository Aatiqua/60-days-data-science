```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("statsfinal.csv")
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
# Handle missing values

for column in df.columns:
    if df[column].isnull().sum() > 0:
        if df[column].dtype == "object":
            df[column] = df[column].fillna(df[column].mode()[0])
        else:
            df[column] = df[column].fillna(df[column].median())

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
# Check number of duplicate rows
print("Duplicates before cleaning:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

print("Duplicates after cleaning:", df.duplicated().sum())
```

```python
print("Data types before cleaning:")
print(df.dtypes)
```

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

print("Data types after cleaning:")
print(df.dtypes)
```

```python
# Feature 1: Total sales across all products

df["Total_Sales"] = df[["S-P1", "S-P2", "S-P3", "S-P4"]].sum(axis=1)

print(df[["Date", "S-P1", "S-P2", "S-P3", "S-P4", "Total_Sales"]].head())
```

```python
# Feature 2: Total quantity of all products

df["Total_Quantity"] = df[["Q-P1", "Q-P2", "Q-P3", "Q-P4"]].sum(axis=1)

print(df[["Date", "Q-P1", "Q-P2", "Q-P3", "Q-P4", "Total_Quantity"]].head())
```

```python
#Save the cleaned dataset
df.to_csv("cleaned_statsfinal.csv",index=False)
print("cleaned dataset saved successfully!")
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)

print("\nNew features:")
print(df[["Total_Sales", "Total_Quantity"]].head())
```

Day 9 — Data Cleaning Decisions

1. Missing Values

Missing values were identified using "df.isnull().sum()".

For numeric columns, missing values were replaced with the median of the respective column. The median was chosen because it is less affected by extreme values than the mean.

For text/categorical columns, missing values were replaced with the mode, which represents the most frequently occurring value.

2. Duplicate Records

Duplicate records were identified using "df.duplicated().sum()".

Duplicate rows were removed using "df.drop_duplicates()" to prevent repeated records from affecting the analysis.

3. Data Format

The "Date" column was converted from text format to a proper datetime format using "pd.to_datetime()".

This makes the date information easier to analyze and prevents inconsistent date handling.

4. New Features

Two new features were created:

- Total_Sales — total sales across Products 1–4.
- Total_Quantity — total quantity across Products 1–4.

5. Final Validation

After cleaning, the dataset was checked again for:

- Missing values
- Duplicate records
- Correct data types
- Newly created features

The cleaned dataset was saved as "cleaned_statsfinal.csv".
