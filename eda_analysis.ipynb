```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("cleaned_statsfinal.csv")

print("Dataset loaded successfully!")
print(df.shape)
```

```python
plt.title("Distribution of Total Sales")
plt.xlabel("Total Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Total_Quantity"], kde=True)

plt.title("Distribution of Total Quantity")
plt.xlabel("Total Quantity")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x=df["Total_Sales"])

plt.title("Outliers in Total Sales")
plt.xlabel("Total Sales")
plt.show()
```

```python
print("Total Sales Statistics:")
print(df["Total_Sales"].describe())

print("\nTotal Quantity Statistics:")
print(df["Total_Quantity"].describe())

print("\nHighest Total Sales:")
print(df["Total_Sales"].max())

print("\nLowest Total Sales:")
print(df["Total_Sales"].min())
```

## 5 Key EDA Insights

1. The dataset contains 4,600 records, providing a substantial number of observations for analysis.

2. The average Total Sales is approximately 51,632.71, while the median is approximately 51,873.19, showing that sales are centered around 52,000.

3. Total Sales range from 11,296.30 to 93,819.41, indicating considerable variation in sales across the dataset.

4. The average Total Quantity is approximately 10,521.37, with values ranging from 2,069 to 19,108.

5. The box plot and distribution plots indicate variation in sales and quantity, with some extreme values that may represent potential outliers and should be investigated further.
