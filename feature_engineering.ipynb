```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
df = pd.read_csv("cleaned_statsfinal.csv")

print("Dataset loaded successfully!")
print(df.head())
```

```python
categorical_features = df.select_dtypes(include=["object"]).columns.tolist()
numerical_features = df.select_dtypes(include=["number"]).columns.tolist()

print("Categorical Features:")
print(categorical_features)

print("\nNumerical Features:")
print(numerical_features)
```

```python
df_encoded = pd.get_dummies(df, columns=categorical_features, drop_first=True)

print("Shape before encoding:", df.shape)
print("Shape after encoding:", df_encoded.shape)
print(df_encoded.head())
```

```python
scaler = StandardScaler()

df_encoded[numerical_features] = scaler.fit_transform(
    df_encoded[numerical_features]
)

print("Numerical features have been scaled.")
print(df_encoded.head())
```

```python
# Create two derived features
df_encoded["Sales_per_Quantity"] = (
    df_encoded["Total_Sales"] / df_encoded["Total_Quantity"]
)

df_encoded["Sales_per_Product"] = df_encoded["Total_Sales"] / 4

print(df_encoded[["Sales_per_Quantity", "Sales_per_Product"]].head())
```

```python
print("BEFORE Feature Engineering")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Missing values:", df.isnull().sum().sum())

print("\nAFTER Feature Engineering")
print("Rows:", df_encoded.shape[0])
print("Columns:", df_encoded.shape[1])
print("Missing values:", df_encoded.isnull().sum().sum())
```

```python
df_encoded.to_csv("feature_engineered_statsfinal.csv", index=False)

print("Feature-engineered dataset saved successfully!")
```

# Day 10 — Feature Engineering

## Objective

The goal of this task is to prepare the cleaned dataset for machine learning by transforming features into a suitable format.

## 1. Feature Identification

The dataset was divided into:

- *Numerical features:* Features containing numerical values.
- *Categorical features:* Features containing text or category values.

Pandas select_dtypes() was used to identify these features.

## 2. Categorical Encoding

Categorical features were converted into numerical form using one-hot encoding.

This allows machine learning algorithms to work with categorical information.

## 3. Numerical Scaling

Numerical features were standardized using StandardScaler.

Scaling puts numerical features on a comparable scale and helps many machine learning algorithms perform more effectively.

## 4. Derived Features

Two new features were created:

- *Sales_per_Quantity:* Total sales divided by total quantity.
- *Sales_per_Product:* Total sales divided by the number of products.

These features provide additional information that can be useful for analysis and machine learning.

## 5. Model Readiness

Before feature engineering, the dataset contained the original numerical and categorical features.

After feature engineering:

- Categorical data was encoded.
- Numerical features were scaled.
- Two additional derived features were created.
- The resulting dataset is more suitable for machine learning.

## 6. Output

The final feature-engineered dataset was saved as:

feature_engineered_statsfinal.csv
